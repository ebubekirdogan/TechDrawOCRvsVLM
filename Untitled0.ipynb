````python
%cd /content/edocr2
import cv2, os, json # cv; goruntu isleme icin
import numpy as np # matematiksel islemler icin
from edocr2 import tools
import tensorflow as tf # derin ogrenme altyapisi
from edocr2.keras_ocr.recognition import Recognizer # yazi okuma islemleri icin
from edocr2.keras_ocr.detection import Detector # !Metin Tespiti islemleri icin Keras-OCR kutuphanesinin tespit edicisi kullanilir.

# --- Resmi yükle ---
file_path = '/content/drive/MyDrive/vlm_ocr_teknik_resim/input/teknik_resim_test.png'
img = cv2.imread(file_path)
filename = os.path.splitext(os.path.basename(file_path))[0]

# --- !Segmentasyon (CV yontemleri ile) ---

#!(On Isleme) binary_thres=127-> 127 den kucuk pikseller siyah, buyuk pikseller beyaz olur. 
# GDT_thres=0.02-> GD&T kutulari icin 0.02 den buyuk olursa GD&T kutusu olarak kabul edilir.
# frame_thres=0.7-> cerceve tespitinde 0.7 den buyuk olursa cerceve olarak kabul edilir.
# autoframe=True-> cerceve tespitinde otomatik olarak cerceveyi bulur ve keser.

img_boxes, frame, gdt_boxes, tables, dim_boxes = tools.layer_segm.segment_img(
    img, autoframe=True, frame_thres=0.7, GDT_thres=0.02, binary_thres=127)

# --- Modelleri yükle ---
gpus = tf.config.list_physical_devices('GPU')
for gpu in gpus:
    tf.config.experimental.set_memory_growth(gpu, True)

# --- gdt-> GD&T ve Ölçüler için -> dim modelleri yükle, onceden egitimli modeller ---
gdt_model = 'edocr2/models/recognizer_gdts.keras'
dim_model = 'edocr2/models/recognizer_dimensions_2.keras'

# Segmentasyon asamasinda GD&T kutuları varsa, GD&T tanıyıcıyı yükle
recognizer_gdt = None
if gdt_boxes:
    recognizer_gdt = Recognizer(alphabet=tools.ocr_pipelines.read_alphabet(gdt_model))
    recognizer_gdt.model.load_weights(gdt_model)

# olculeri okumak icin modeli yukle
alphabet_dim = tools.ocr_pipelines.read_alphabet(dim_model)
recognizer_dim = Recognizer(alphabet=alphabet_dim)
recognizer_dim.model.load_weights(dim_model)

# Yazinin nerede oldugunu bulmak icin detector yukle
detector = Detector()

# --- OCR: Tablo (Tesseract) ---
process_img = img.copy() # orijinal resmi bozmamak icin kopya alinir
table_results, updated_tables, process_img = tools.ocr_pipelines.ocr_tables(tables, process_img, 'eng')

# --- OCR: GD&T ---
gdt_results, updated_gdt_boxes, process_img = tools.ocr_pipelines.ocr_gdt(process_img, gdt_boxes, recognizer_gdt)

# --- resmin etrafinda bir teknik resim cercevesi bulunmussa resmi o cercevenin icinden kesip cikarir ---
if frame:
    process_img = process_img[frame.y : frame.y + frame.h, frame.x : frame.x + frame.w]
dimensions, other_info, process_img, dim_tess = tools.ocr_pipelines.ocr_dimensions(
    process_img, detector, recognizer_dim, alphabet_dim, frame, dim_boxes,
    cluster_thres=20, max_img_size=1048, language='eng', backg_save=False) # cluster_thres=20-> 20 pikselden daha yakin olan boyutlar birlestirilir. ornegin bir ölçü Ø 50 H7 ise bunlari ayri ayri degil birlikte isle.
                                                                           # max_img_size=1048-> resim boyutu 1048 den buyukse kucultup modele oyle ver.

# --- Çıktıyı işle ve Drive'a kaydet ---
output_path = f'/content/{filename}'
table_results, gdt_results, dimensions, other_info = tools.output_tools.process_raw_output(
    output_path, table_results, gdt_results, dimensions, other_info, save=False)

result = {'tables': table_results, 'gdt': gdt_results, 'dimensions': dimensions, 'other_info': other_info}

drive_output = '/content/drive/MyDrive/vlm_ocr_teknik_resim/edocr2_output/edocr2_output.json'
with open(drive_output, 'w', encoding='utf-8') as f:
    json.dump(result, f, ensure_ascii=False, indent=2, default=str)

print("Kaydedildi:", drive_output)
print(json.dumps(result, indent=2, default=str, ensure_ascii=False))
````

````python
# API anahtarını Google Colab kullanıcı verilerinden al
from google.colab import userdata
from google import genai

api_key = userdata.get('GEMINI_API_KEY')
client = genai.Client(api_key=api_key)
````

````python
import time, json, re
from PIL import Image

img_path = '/content/drive/MyDrive/vlm_ocr_teknik_resim/input/teknik_resim_test.png'
image = Image.open(img_path)

prompt = """
You are analyzing a mechanical engineering technical drawing.
Extract ALL manufacturing information you can find in this drawing and return it as a structured JSON object.

Include (if present):
- All dimensions with their tolerances (diameters, lengths, depths, angles, radii)
- All GD&T (Geometric Dimensioning and Tolerancing) callouts with their symbols, values, and datum references
- Surface roughness (Ra) values
- Material
- Title block info (part name, drawing number, revision, scale, projection)
- General tolerance notes
- Thread specifications
- Any other manufacturing notes

Return ONLY a valid JSON object, no explanation text before or after.
"""

max_retries = 5
start_time = time.time()
response = None # modelden gelecek olan yaniti tutacak olan degisken

for attempt in range(max_retries):
    try:
        response = client.models.generate_content(
            model='gemini-3.6-flash',
            contents=[image, prompt]
        )
        break
    except Exception as e:
        print(f"Deneme {attempt+1} başarısız: {e}")
        if attempt < max_retries - 1:
            wait = 15 * (attempt + 1)
            print(f"{wait} saniye bekleyip tekrar denenecek...")
            time.sleep(wait)
        else:
            raise

elapsed = time.time() - start_time
print(f"Süre: {elapsed:.2f} saniye")
print(response.text)
````

````python
import re

raw_text = response.text.strip()
cleaned = re.sub(r'^```json\s*|\s*```$', '', raw_text, flags=re.MULTILINE).strip()

try:
    gemini_json = json.loads(cleaned)
except json.JSONDecodeError:
    print("UYARI: JSON parse edilemedi, ham metin kaydediliyor")
    gemini_json = {"raw_output": cleaned}

result = {
    "output": gemini_json,
    "processing_time_seconds": elapsed
}

drive_output = '/content/drive/MyDrive/vlm_ocr_teknik_resim/gemini_output/gemini_output.json'
with open(drive_output, 'w', encoding='utf-8') as f:
    json.dump(result, f, ensure_ascii=False, indent=2)

print("Kaydedildi:", drive_output)
print(json.dumps(result, indent=2, ensure_ascii=False))
````
